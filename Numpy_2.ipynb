### Problem 2 - Slicing

```python
arr = np.array([10, 20, 30, 40, 50, 60, 70, 80, 90])
```

```python
# A. First 5 elements
print(arr[:5])
```

```python
# B. Last 4 elements
print(arr[-4:])
```

```python
# C. Elements from index 2 to index 6
print(arr[2:7])
```

```python
# D. Every second element
print(arr[::2])
```

```python
# E. Reverse the entire array
print(arr[::-1])
```

### Problem 3 - Boolean Masking

```python
import numpy as np
marks = np.array([45, 67, 82, 39, 91, 56, 74, 88, 32, 95])
```

```python
# 1. All marks greater than 70
above_70 = marks[marks>70]
print(above_70)
```

```python
# 2. All marks less than 50
below_50 = marks[marks<50]
print(below_50)
```

```python
# 3. All marks greater than or equal to 80
above_80 = marks[marks>=80]
print(above_80)
```

```python
# 4. Number of students who scored above 70
count_above_70 = np.sum(marks>70)
print(count_above_70)
```

```python
# 5. Number of students who failed (marks < 40)
failed_students = np.sum(marks<40)
print(failed_students)
```

### Problem 5 - Student Performance Analysis

```python
marks = np.array([45, 78, 92, 56, 34, 88, 67, 95, 41, 73])
```

```python
# 1. Passing students
passing_students = marks[marks>=50]
print(passing_students)
```

```python
# 2. Failing students
failing_marks = marks[marks<50]
print(failing_marks)
```

```python
# 3. Count passing students
count_passing_student = np.sum(marks>=50)
print(count_passing_student)
```

```python
# 4. Count failing students
count_failing_student = np.sum(marks<50)
print(count_failing_student)
```

```python
# 5. Find the highest marks among passing students
highest_passing = np.max(marks[marks >= 50])
print("Highest passing mark:", highest_passing)
```
