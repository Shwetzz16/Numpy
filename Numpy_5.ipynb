### Problem 1 - Scalar Broadcasting

```python
import numpy as np
prices = np.array([100, 200, 300, 400, 500])
```

```python
# 1. New prices
new_prices = prices + 50
print("New prices:", new_prices)
```

```python
# 2. Average
average = np.mean(new_prices)
print("Average:", average)
```

```python
# 3. Difference
difference = np.max(new_prices) - np.min(new_prices)
print("Difference:", difference)
```

### Problem 2 - Apply Discount

```python
import numpy as np
products = np.array([1000, 2000, 3000, 4000, 5000])
```

```python
# 1. Discount amount
discount = products * 0.10
print("Discount:", discount)
```

```python
# 2. Final prices
final_price = products - discount
print("Final prices:", final_price)
```

```python
# 3. Total original price
total_original = np.sum(products)
print("Total original:", total_original)
```

```python
# 4. Total after discount
total_after_discount = np.sum(final_price)
print("Total after discount:", total_after_discount)
```

### Problem 3 - Matrix + Vector Broadcasting

```python
import numpy as np

marks = np.array([
    [70, 80, 90],
    [60, 75, 85],
    [88, 92, 79]
])

bonus = np.array([5, 3, 2])
```

```python
# 1. Add bonus marks
new_marks = marks + bonus
print("New marks:")
print(new_marks)
```

```python
# 2. Total marks for each student
total_marks = np.sum(new_marks, axis=1)
print("Total marks:", total_marks)
```

```python
# 3. Average marks for each student
average_marks = np.mean(new_marks, axis=1)
print("Average marks:", np.round(average_marks, 2))
```

```python
# 4. Highest mark
highest_mark = np.max(new_marks)
print("Highest mark:", highest_mark)
```

### Problem 5 : Real-World Employee Salary Analysis

```python
import numpy as np

salaries = np.array([
    30000, 35000, 40000,
    45000, 50000, 55000,
    60000, 65000, 70000, 75000
])
```

```python
# 1. Raise amount
raise_amount = salaries * 0.12
print("Raise amount:", raise_amount)
```

```python
# 2. New salaries
new_salaries = salaries + raise_amount
print("New salaries:", new_salaries)
```

```python
# 3. Average before raise
average_before = np.mean(salaries)
print("Average before:", average_before)
```

```python
# 4. Average after raise
average_after = np.mean(new_salaries)
print("Average after:", average_after)
```

```python
# 5. Total additional expense
total_additional = np.sum(raise_amount)
print("Total additional expense:", total_additional)
```

```python
# 6. Salaries above ₹60,000
above_60000 = new_salaries[new_salaries > 60000]
print("Salaries above ₹60,000:", above_60000)
```

```python
# 7. Count employees above ₹60,000
count_above_60000 = np.sum(new_salaries > 60000)
print("Number of employees:", count_above_60000)
```

```python
# 8. Percentage increase
total_old = np.sum(salaries)
total_new = np.sum(new_salaries)

percentage_increase = ((total_new - total_old) / total_old) * 100
print("Percentage increase:", percentage_increase)
```
