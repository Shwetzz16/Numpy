### Problem 1 - Basic Statistics

```python
import numpy as np

salary = np.array([
    35000, 42000, 50000, 38000,
    47000, 55000, 62000, 45000
])
```

```python
# 1. Average salary
average = np.mean(salary)
print("Average salary:", average)
```

```python
# 2. Median salary
median = np.median(salary)
print("Median salary:", median)
```

```python
# 3. Minimum salary
minimum = np.min(salary)
print("Minimum salary:", minimum)
```

```python
# 4. Maximum salary
maximum = np.max(salary)
print("Maximum salary:", maximum)
```

```python
# 5. Total salary paid
total = np.sum(salary)
print("Total salary:", total)
```

### Problem 2 - Standard Deviation & Variance

```python
import numpy as np

scores = np.array([
    72, 75, 78, 80, 82,
    85, 88, 90, 92, 95
])
```

```python
# 1. Mean
mean = np.mean(scores)
print("Mean:", mean)
```

```python
# 2. Median
median = np.median(scores)
print("Median:", median)
```

```python
# 3. Standard deviation
std = np.std(scores)
print("Standard deviation:", std)
```

```python
# 4. Variance
variance = np.var(scores)
print("Variance:", variance)
```

```python
# 5. Difference between maximum and minimum
range_value = np.max(scores) - np.min(scores)
print("Range:", range_value)
```

### Problem 4 - Min-Max Normalization

```python
import numpy as np
data = np.array([10, 20, 30, 40, 50])
```

```python
# 1. Find the minimum
minimum = np.min(data)
print("Minimum:", minimum)
```

```python
# 2. Find the maximum
maximum = np.max(data)
print("Maximum:", maximum)
```

```python
# 3. Calculate the normalized array
normalized = (data - np.min(data)) / (np.max(data) - np.min(data))
print("Normalized data:", normalized)
```

```python
# 4. Verify minimum normalized value is 0
print("Minimum normalized value:", np.min(normalized))
print("Minimum is 0:", np.min(normalized) == 0)
```

```python
# 5. Verify maximum normalized value is 1
print("Maximum normalized value:", np.max(normalized))
print("Maximum is 1:", np.max(normalized) == 1)
```

### Problem 5 - Real-World Sales Statistics Challenge

```python
import numpy as np

sales = np.array([
    12000, 15000, 18000, 14000,
    21000, 25000, 22000, 28000,
    30000, 27000, 32000, 35000
])
```

```python
# 1. Total annual sales
total_sales = np.sum(sales)
print("Total annual sales:", total_sales)
```

```python
# 2. Average monthly sales
average_sales = np.mean(sales)
print("Average monthly sales:", average_sales)
```

```python
# 3. Median monthly sales
median_sales = np.median(sales)
print("Median monthly sales:", median_sales)
```

```python
# 4. Month with highest sales
highest_index = np.argmax(sales)
print("Highest sales month:", highest_index + 1)
print("Highest sales:", np.max(sales))
```

```python
# 5. Months above average
months_above_average = np.sum(sales > average_sales)
print("Months above average:", months_above_average)
```

```python
# 6. Standard deviation
std_sales = np.std(sales)
print("Standard deviation:", std_sales)
```

```python
# 7. Percentage of months above ₹25,000
count = np.sum(sales > 25000)
percentage = (count / sales.size) * 100
print("Percentage above ₹25,000:", percentage)
```
