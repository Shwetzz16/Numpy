### Problem 1 - Student Performance Analysis

```python
import numpy as np

marks = np.array([
    [78, 85, 92],
    [65, 70, 68],
    [90, 88, 95],
    [55, 60, 58],
    [82, 79, 85],
    [45, 50, 48],
    [92, 95, 89],
    [70, 75, 72]
])
```

```python
# 1. Total marks for each student
student_total = np.sum(marks, axis=1)
print("Total marks:", student_total)
```

```python
# 2. Average marks for each student
student_average = np.mean(marks, axis=1)
print("Average marks:", student_average)
```

```python
# 3. Highest total score
highest_total = np.max(student_total)
print("Highest total score:", highest_total)
```

```python
# 4. Student index with the highest total score
highest_index = np.argmax(student_total)
print("Student index:", highest_index)
```

```python
# 5. Students whose average is above 75
above_75 = student_average[student_average > 75]
print("Averages above 75:", above_75)
```

### Problem 2 - Employee Salary Analysis

```python
import numpy as np

salaries = np.array([
    35000, 42000, 58000, 62000,
    48000, 75000, 82000, 55000,
    68000, 91000
])
```

```python
# 1. Average salary
average = np.mean(salaries)
print("Average salary:", average)
```

```python
# 2. Median salary
median = np.median(salaries)
print("Median salary:", median)
```

```python
# 3. Minimum salary
minimum = np.min(salaries)
print("Minimum salary:", minimum)
```

```python
# 4. Maximum salary
maximum = np.max(salaries)
print("Maximum salary:", maximum)
```

```python
# 5. Employees earning above ₹60,000
above_60000 = salaries[salaries > 60000]
print("Salaries above ₹60,000:", above_60000)
```

```python
# 6. Number of employees earning above ₹60,000
count = np.sum(salaries > 60000)
print("Number of employees:", count)
```

```python
# 7. Percentage of employees earning above ₹60,000
percentage = (count / salaries.size) * 100
print("Percentage:", percentage)
```

```python
# 8. Categorize salaries as "High" or "Normal"
category = np.where(
    salaries >= 60000,
    "High",
    "Normal"
)

print("Salary category:", category)
```

### Problem 3 - Monthly Sales Analysis

```python
import numpy as np

sales = np.array([
    15000, 18000, 22000, 17000,
    25000, 28000, 24000, 30000,
    32000, 29000, 35000, 38000
])
```

```python
# 1. Total annual sales
total_sales = np.sum(sales)
print("Total annual sales:", total_sales)
```

```python
# 2. Average monthly sales
average_sales = np.mean(sales)
print("Average monthly sales:", average_sales)
```

```python
# 3. Month with the highest sales
highest_index = np.argmax(sales)

print("Highest sales index:", highest_index)
print("Highest sales month:", highest_index + 1)
print("Highest sales:", np.max(sales))
```

```python
# 4. Month with the lowest sales
lowest_index = np.argmin(sales)

print("Lowest sales month:", lowest_index + 1)
print("Lowest sales:", np.min(sales))
```

```python
# 5. Number of months where sales exceeded ₹25,000
count = np.sum(sales > 25000)
print("Number of months:", count)
```

```python
# 6. All months/sales where sales exceeded ₹25,000
above_25000 = sales[sales > 25000]
print("Sales above ₹25,000:", above_25000)
```

```python
# 7. Percentage of months where sales exceeded ₹25,000
percentage = (count / sales.size) * 100
print("Percentage:", percentage)
```

```python
# 8. Sales growth from January to December
january = sales[0]
december = sales[-1]

growth = ((december - january) / january) * 100

print("Sales growth:", growth, "%")
```

### Problem 4 - Temperature Analysis

```python
import numpy as np

temperature = np.array([
    28, 31, 35, 33, 29,
    37, 39, 34, 30, 32
])
```

```python
# 1. Average temperature
average = np.mean(temperature)
print("Average temperature:", average)
```

```python
# 2. Highest temperature
highest = np.max(temperature)
print("Highest temperature:", highest)
```

```python
# 3. Lowest temperature
lowest = np.min(temperature)
print("Lowest temperature:", lowest)
```

```python
# 4. Number of days above 35°C
days_above_35 = np.sum(temperature > 35)
print("Days above 35°C:", days_above_35)
```

```python
# 5. Number of days below 30°C
days_below_30 = np.sum(temperature < 30)
print("Days below 30°C:", days_below_30)
```

```python
# 6. Find all temperatures above 35°C
hot_temperatures = temperature[temperature > 35]
print("Temperatures above 35°C:", hot_temperatures)
```

```python
# 7. Replace temperatures above 35°C with "Hot"
temperature_status = np.where(
    temperature > 35,
    "Hot",
    "Normal"
)

print(temperature_status)
```

```python
# 8. Find the temperature range
temperature_range = np.max(temperature) - np.min(temperature)
print("Temperature range:", temperature_range)
```

### Problem 5 - Real-World E-Commerce Analysis

```python
import numpy as np

orders = np.array([
    1200, 2500, 800, 4500, 3200,
    1500, 6000, 900, 2800, 5200
])

ratings = np.array([
    4.5, 4.2, 3.8, 4.8, 4.1,
    3.5, 4.9, 3.9, 4.3, 4.7
])
```

```python
# 1. Total revenue
total_revenue = np.sum(orders)
print("Total revenue:", total_revenue)
```

```python
# 2. Average order value
average_order = np.mean(orders)
print("Average order value:", average_order)
```

```python
# 3. Highest order value
highest_order = np.max(orders)
print("Highest order:", highest_order)
```

```python
# 4. Lowest order value
lowest_order = np.min(orders)
print("Lowest order:", lowest_order)
```

```python
# 5. All orders above ₹3,000
orders_above_3000 = orders[orders > 3000]
print("Orders above ₹3,000:", orders_above_3000)
```

```python
# 6. Count orders above ₹3,000
count_above_3000 = np.sum(orders > 3000)
print("Number of orders:", count_above_3000)
```

```python
# 7. Percentage of orders above ₹3,000
percentage = (count_above_3000 / orders.size) * 100
print("Percentage:", percentage)
```

```python
# 8. Average customer rating
average_rating = np.mean(ratings)
print("Average rating:", average_rating)
```

```python
# 9. Number of customers with rating ≥ 4.5
high_rating_count = np.sum(ratings >= 4.5)
print("Customers with rating >= 4.5:", high_rating_count)
```

```python
# 10. Order amount corresponding to the highest rating
highest_rating_index = np.argmax(ratings)
print("Highest rating index:", highest_rating_index)
```
